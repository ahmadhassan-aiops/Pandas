# 1. Opening a csv file from an URL

```python
# mostly use for big data (you just need to change the url for your data source)
import requests
import pandas as pd
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data).head()
```

# 9. Skip bad lines

```python
# Simulating a CSV file with some bad lines
data = """ISBN;Book-Title;Book-Author;Year-Of-Publication;Publisher
1234567890;The Great Gatsby;F. Scott Fitzgerald;1925;Scribner
9876543210;To Kill a Mockingbird;Harper Lee;1960;J.B. Lippincott & Co.
BAD LINE HERE
5432167890;1984;George Orwell;1949
6543219870;Pride and Prejudice;Jane Austen;1813;T. Egerton
8765432109;Moby Dick;Herman Melville;;Harper & Brothers
EXTRA;COLUMN;HERE;WRONG;DATA;TOO
"""

# Use StringIO to simulate reading from a file
csv_file = StringIO(data)

# Step 2: Load the DataFrame and correct it by skipping bad lines
df = pd.read_csv(csv_file, sep=';', encoding="latin-1", on_bad_lines='skip')

# Display the corrected DataFrame
display(df)
```

# 12. Convertors


```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name

pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```

# 14. Loading a huge dataset in chunks


**Consideration here is that every action will be performed in loops here!**

```python
dfs = pd.read_csv('aug_train.csv',chunksize=5000)
for chunks in dfs:
    print(chunks.shape)
```
